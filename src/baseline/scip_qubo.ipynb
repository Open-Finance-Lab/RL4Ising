````python
from pyscipopt import Model
````

````python
model = Model()
x = model.addVar(vtype="B")
y = model.addVar(vtype="B")

model.addCons(x + y <= 1)
model.setObjective(2 * x + y, "maximize")
model.optimize()

print("SCIP is working:", model.getStatus() == "optimal" and model.getObjVal() == 2)
````
